# 🔍 2. Data Validation | Validação dos Dados

## 🇧🇷 Português

Validação rápida da tabela analítica antes da criação das consultas utilizadas no Dashboard.

---

## 🇺🇸 English

Quick validation of the analytical table before creating the queries used by the Dashboard.

```python
%sql
SELECT
    COUNT(*) AS total_rows,
    COUNT(DISTINCT asin) AS unique_products,
    COUNT(DISTINCT category_name) AS total_categories
FROM gold.product_analysis;
```

# 🎯 3. KPI Overview | Visão Geral dos KPIs

## 🇧🇷 Português

Consulta principal para os indicadores de desempenho exibidos no topo do Dashboard.

---

## 🇺🇸 English

Main query for the performance indicators displayed at the top of the Dashboard.

```python
%sql
SELECT
    COUNT(*) AS total_products,
    COUNT(DISTINCT category_name) AS total_categories,
    ROUND(AVG(price), 2) AS average_price,
    ROUND(AVG(discount_percentage), 2) AS average_discount,
    ROUND(AVG(stars), 2) AS average_rating,
    SUM(boughtInLastMonth) AS total_bought
FROM gold.product_analysis;
```

```python

%md
# 📦 4. Category Performance | Desempenho por Categoria

## 🇧🇷 Português

Comparação das categorias por quantidade de produtos, preço, desconto, avaliação e volume de compras.

---

## 🇺🇸 English

Comparison of categories by product count, price, discount, rating and purchase volume.
```

```python
%sql
SELECT
    category_name,
    COUNT(*) AS total_products,
    ROUND(AVG(price), 2) AS average_price,
    ROUND(AVG(discount_percentage), 2) AS average_discount,
    ROUND(AVG(stars), 2) AS average_rating,
    SUM(boughtInLastMonth) AS total_bought
FROM gold.product_analysis
GROUP BY category_name
HAVING COUNT(*) >= 100
ORDER BY total_bought DESC
LIMIT 20;
```

# 💰 5. Price Range Analysis | Análise por Faixa de Preço

## 🇧🇷 Português

Distribuição dos produtos por faixa de preço e comparação com o volume de compras.

---

## 🇺🇸 English

Product distribution by price range and comparison with purchase volume.


```python
%sql
SELECT
    price_range,
    COUNT(*) AS total_products,
    ROUND(AVG(price), 2) AS average_price,
    ROUND(AVG(discount_percentage), 2) AS average_discount,
    SUM(boughtInLastMonth) AS total_bought
FROM gold.product_analysis
GROUP BY price_range
ORDER BY total_products DESC;
```

# ⭐ 6. Rating Distribution | Distribuição das Avaliações

## 🇧🇷 Português

Distribuição dos produtos por categoria de avaliação para identificar a qualidade percebida do catálogo.

---

## 🇺🇸 English

Distribution of products by rating category to identify perceived catalog quality.


```python
%sql
SELECT
    rating_category,
    COUNT(*) AS total_products,
    ROUND(
        COUNT(*) * 100.0 /
        SUM(COUNT(*)) OVER (),
        2
    ) AS percentage
FROM gold.product_analysis
GROUP BY rating_category
ORDER BY total_products DESC;
```

# 🏆 7. Best Sellers | Best Sellers

## 🇧🇷 Português

Comparação entre Best Sellers e os demais produtos considerando preço, avaliação, reviews e desconto.

---

## 🇺🇸 English

Comparison between Best Sellers and other products based on price, rating, reviews and discount.

```python
%sql
SELECT
    best_seller_flag,
    COUNT(*) AS total_products,
    ROUND(AVG(stars), 2) AS average_rating,
    ROUND(AVG(reviews), 2) AS average_reviews,
    ROUND(AVG(price), 2) AS average_price,
    ROUND(AVG(discount_percentage), 2) AS average_discount,
    SUM(boughtInLastMonth) AS total_bought
FROM gold.product_analysis
GROUP BY best_seller_flag
ORDER BY total_bought DESC;
```

# 🔥 8. Most Popular Products | Produtos Mais Populares

## 🇧🇷 Português

Ranking dos produtos com maior volume de compras no último mês.

---

## 🇺🇸 English

Ranking of products with the highest purchase volume in the last month.


```python
%sql
SELECT
    asin,
    title,
    category_name,
    price,
    stars,
    reviews,
    boughtInLastMonth
FROM gold.product_analysis
ORDER BY boughtInLastMonth DESC
LIMIT 20;
```

# 💸 9. Discount Analysis | Análise de Descontos

## 🇧🇷 Português

Identificação das maiores oportunidades de desconto presentes no catálogo.

---

## 🇺🇸 English

Identification of the largest discount opportunities in the catalog.


```python
%sql
SELECT
    price_range,
    COUNT(*) AS total_products,
    ROUND(AVG(discount_percentage), 2) AS average_discount,
    ROUND(MAX(discount_percentage), 2) AS maximum_discount
FROM gold.product_analysis
GROUP BY price_range
ORDER BY average_discount DESC;
```

# 🥇 10. Category Ranking | Ranking de Categorias

## 🇧🇷 Português

Ranking das categorias por volume de compras utilizando Window Functions.

---

## 🇺🇸 English

Category ranking by purchase volume using Window Functions.


```python
%sql
WITH category_metrics AS (
    SELECT
        category_name,
        COUNT(*) AS total_products,
        ROUND(AVG(stars), 2) AS average_rating,
        ROUND(AVG(discount_percentage), 2) AS average_discount,
        SUM(boughtInLastMonth) AS total_bought
    FROM gold.product_analysis
    GROUP BY category_name
    HAVING COUNT(*) >= 100
)
SELECT
    category_name,
    total_products,
    average_rating,
    average_discount,
    total_bought,
    RANK() OVER (
        ORDER BY total_bought DESC
    ) AS purchase_rank
FROM category_metrics
ORDER BY purchase_rank
LIMIT 20;
```

# 📊 11. Dashboard Views | Views para o Dashboard

## 🇧🇷 Português

Criação de views temporárias para organizar as principais fontes de dados do Dashboard.

---

## 🇺🇸 English

Creation of temporary views to organize the main Dashboard data sources.


```python
%sql
CREATE OR REPLACE VIEW gold.vw_dashboard_kpis AS
SELECT
    COUNT(*) AS total_products,
    COUNT(DISTINCT category_name) AS total_categories,
    ROUND(AVG(price), 2) AS average_price,
    ROUND(AVG(discount_percentage), 2) AS average_discount,
    ROUND(AVG(stars), 2) AS average_rating,
    SUM(boughtInLastMonth) AS total_bought
FROM gold.product_analysis;
```

```python
%sql
CREATE OR REPLACE VIEW gold.vw_dashboard_categories AS
SELECT
    category_name,
    COUNT(*) AS total_products,
    ROUND(AVG(price), 2) AS average_price,
    ROUND(AVG(discount_percentage), 2) AS average_discount,
    ROUND(AVG(stars), 2) AS average_rating,
    SUM(boughtInLastMonth) AS total_bought
FROM gold.product_analysis
GROUP BY category_name
HAVING COUNT(*) >= 100;
```

```python
%sql
CREATE OR REPLACE VIEW gold.vw_dashboard_price_range AS
SELECT
    price_range,
    COUNT(*) AS total_products,
    ROUND(AVG(price), 2) AS average_price,
    ROUND(AVG(discount_percentage), 2) AS average_discount,
    SUM(boughtInLastMonth) AS total_bought
FROM gold.product_analysis
GROUP BY price_range;
```

```python
%sql
CREATE OR REPLACE VIEW gold.vw_dashboard_rating AS
SELECT
    rating_category,
    COUNT(*) AS total_products,
    ROUND(AVG(stars), 2) AS average_rating
FROM gold.product_analysis
GROUP BY rating_category;
```

```python
%sql
CREATE OR REPLACE VIEW gold.vw_dashboard_best_sellers AS
SELECT
    best_seller_flag,
    COUNT(*) AS total_products,
    ROUND(AVG(stars), 2) AS average_rating,
    ROUND(AVG(reviews), 2) AS average_reviews,
    ROUND(AVG(price), 2) AS average_price,
    ROUND(AVG(discount_percentage), 2) AS average_discount,
    SUM(boughtInLastMonth) AS total_bought
FROM gold.product_analysis
GROUP BY best_seller_flag;
```
