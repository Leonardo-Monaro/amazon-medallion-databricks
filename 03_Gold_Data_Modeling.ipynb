```python
%sql
CREATE SCHEMA IF NOT EXISTS gold;
```

```python
silver = spark.table("silver.products")

product_analysis = silver.select(
    "asin",
    "title",
    "category_name",
    "price",
    "listPrice",
    "discount_amount",
    "discount_percentage",
    "stars",
    "reviews",
    "weighted_rating",
    "rating_category",
    "price_range",
    "best_seller_flag",
    "boughtInLastMonth"
)
```

```python
(
    product_analysis.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable("gold.product_analysis")
)
```

```python
from pyspark.sql.functions import avg, count, sum, round, when, col

category_summary = (
    silver
    .groupBy("category_name")
    .agg(
        count("*").alias("total_products"),
        round(avg("price"),2).alias("avg_price"),
        round(avg("discount_percentage"),2).alias("avg_discount"),
        round(avg("stars"),2).alias("avg_rating"),
        sum("boughtInLastMonth").alias("total_bought")
    )
)
```

```python
category_summary.write \
.format("delta") \
.mode("overwrite") \
.saveAsTable("gold.category_summary")
```

```python
best_sellers = (

    silver

    .filter("best_seller_flag='Yes'")

)
```

```python
best_sellers.write \
.format("delta") \
.mode("overwrite") \
.saveAsTable("gold.best_sellers")
```

```python
discount_analysis = (

    silver

    .groupBy("price_range")

    .agg(

        count("*").alias("products"),

        sum(when(col("is_discounted") == "Yes", 1).otherwise(0)).alias("discounted_products"),

        round(avg("discount_percentage"),2).alias("avg_discount"),

        round(avg(when(col("is_discounted") == "Yes", col("discount_percentage"))),2).alias("avg_discount_when_discounted"),

        round(avg("price"),2).alias("avg_price")

    )

)
```

```python
discount_analysis.write \
.format("delta") \
.mode("overwrite") \
.saveAsTable("gold.discount_analysis")
```
