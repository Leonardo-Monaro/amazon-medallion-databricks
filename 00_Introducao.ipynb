## 🇧🇷 Português

### 📌 Sobre o Projeto

Este projeto tem como objetivo desenvolver um pipeline completo de Engenharia de Dados utilizando a **Arquitetura Medallion** no **Databricks**, aplicando boas práticas de ETL/ELT, modelagem de dados e analytics.

Toda a solução foi construída utilizando **Apache Spark (PySpark)**, **SQL** e **Delta Lake**, simulando um ambiente corporativo de processamento de dados.

O pipeline foi dividido em camadas para garantir escalabilidade, organização e qualidade dos dados.

### 🏗 Arquitetura

```text
Source files / Arquivos de origem
     │
     ▼
🥉 Bronze
Raw Data Ingestion
     │
     ▼
🥈 Silver
Data Cleaning & Business Rules
     │
     ▼
🥇 Gold
Business Data Modeling
     │
     ▼
📊 SQL Analytics
Business Insights
     │
     ▼
📈 Dashboard
Business Intelligence
```

## About the Project

This project implements a complete **Data Engineering Pipeline** using the **Medallion Architecture** within **Databricks**.

The solution was built with **Apache Spark (PySpark)**, **SQL**, and **Delta Lake**, following industry best practices for ETL, data modeling and analytics.

### Architecture

```text
Source files / Arquivos de origem
     │
     ▼
Bronze Layer
     │
     ▼
Silver Layer
     │
     ▼
Gold Layer
     │
     ▼
SQL Analytics
     │
     ▼
Dashboards
```
